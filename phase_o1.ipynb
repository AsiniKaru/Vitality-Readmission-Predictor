```python
# Import required libraries for data analysis
import pandas as pd
import numpy as np
```

```python
# Load the diabetic dataset into a Pandas DataFrame
df = pd.read_csv("data_files/diabetic_data.csv")
```

```python
# Display all column names in the dataset
print("Dataset Columns:")
print(df.columns.tolist())

# Display the total number of columns
print("\nTotal No of Columns:", len(df.columns))
```

```python
# Display the original data types of key ID columns
print("Original Data Types for IDs:")
print(df[["admission_type_id", "discharge_disposition_id", "admission_source_id"]].dtypes)

# Define ID columns that should be treated as categorical variables
categorical_id_columns = [
    "admission_type_id",
    "discharge_disposition_id",
    "admission_source_id"
]

# Convert the selected ID columns into categorical data types
for col in categorical_id_columns:
    df[col] = df[col].astype("category")

# Verify the updated data types
print("\nUpdated Data Types:")
print(df[categorical_id_columns].dtypes)
```

```python
# Define the 22 oral medication columns referenced in the data dictionary
oral_medication_columns = [
    'metformin', 'repaglinide', 'nateglinide', 'chlorpropamide', 
    'glimepiride', 'acetohexamide', 'glipizide', 'glyburide', 
    'tolbutamide', 'pioglitazone', 'rosiglitazone', 'acarbose', 
    'miglitol', 'troglitazone', 'tolazamide', 'examide', 
    'sitagliptin', 'insulin', 'glyburide-metformin', 'glipizide-metformin', 
    'glimepiride-pioglitazone', 'metformin-rosiglitazone', 'metformin-pioglitazone'
]

# Filter and extract available medication columns present in the dataset
available_medications = [col for col in oral_medication_columns if col in df.columns]

# Identify any missing medication columns from the predefined list
missing_medications = [col for col in oral_medication_columns if col not in df.columns]

# Display the verification results
print("--- Oral Medication Cohort Verification ---")
print(f"Total Available Medication Columns: {len(available_medications)}")
print("Available Columns:", available_medications)

print(f"\nTotal Missing Medication Columns: {len(missing_medications)}")
if missing_medications:
    print("Missing Columns:", missing_medications)
else:
    print("Missing Columns: None (All matched successfully!)")
```

```python
# Display the first five records of the available oral medication columns
df[available_medications].head()
```

```python
# Standardize all legacy '?' missing indicators to standard NumPy NaN values
df.replace('?', np.nan, inplace=True)

print("-> Successfully replaced all '?' characters with NumPy NaN values across the DataFrame.")
```

```python
# Investigate the missingness percentage of the 'weight' column
if 'weight' in df.columns:
    missing_weight_count = df['weight'].isnull().sum()
    total_rows = len(df)
    missing_weight_pct = (missing_weight_count / total_rows) * 100
    
    print("--- Weight Column Missingness Audit ---")
    print(f"Total rows: {total_rows}")
    print(f"Missing weight values: {missing_weight_count}")
    print(f"Missingness Percentage: {missing_weight_pct:.2f}%")
else:
    print("-> The 'weight' column is not found in the DataFrame.")
```

```python
# Drop the 'weight' column if missingness exceeds 90% as a Data Quality Limitation
if 'weight' in df.columns:
    missing_weight_pct = (df['weight'].isnull().sum() / len(df)) * 100
    
    if missing_weight_pct > 90.0:
        df.drop(columns=['weight'], inplace=True)
        print("-> Decision: 'weight' column dropped entirely because missingness exceeds 90%.")
        print("\n--- Documentation Note for Report ---")
        print("[Data Quality Limitation]: The 'weight' column was excluded from the analysis ")
        print("due to excessive missingness (>90%) characteristic of legacy clinical record systems.")
    else:
        print("-> Decision: 'weight' column retained (missingness is 90% or below).")
else:
    print("-> 'weight' column has already been dropped.")

# Display the updated dataset shape
print(f"\nUpdated Dataset Shape: {df.shape}")
```
